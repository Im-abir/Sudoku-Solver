```python
sample_grid = [
    [5,3,0, 0,7,0, 0,0,0],
    [6,0,0, 1,9,5, 0,0,0],
    [0,9,8, 0,0,0, 0,6,0],

    [8,0,0, 0,6,0, 0,0,3],
    [4,0,0, 8,0,3, 0,0,1],
    [7,0,0, 0,2,0, 0,0,6],

    [0,6,0, 0,0,0, 2,8,0],
    [0,0,0, 4,1,9, 0,0,5],
    [0,0,0, 0,8,0, 0,7,9]
]

def print_grid(grid):
    for row in grid:
        print(row)
```

```python
import copy
test_grid = copy.deepcopy(sample_grid)

print("Before:")
print_grid(test_grid)

if solve(test_grid):
    print("\nSolved:")
    print_grid(test_grid)
else:
    print("No solution exists")
```

```python
invalid_grid = copy.deepcopy(sample_grid)
invalid_grid[0][0] = 5
invalid_grid[0][1] = 5  # duplicate 5 in same row — makes it unsolvable

if solve(invalid_grid):
    print("Solved (unexpected)")
else:
    print("Correctly detected: no solution exists")
```

```python
import time

test_grid2 = copy.deepcopy(sample_grid)
start = time.time()
solve(test_grid2)
end = time.time()
print(f"Solved in {end - start:.4f} seconds")
```
